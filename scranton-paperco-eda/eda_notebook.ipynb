### 2. Connecting & extracting data
More efficient code -- using `with` context manager.

```python
conn_string = "Driver={SQL Server}; Server=DESKTOP-K5MRD4V\SQLEXPRESS; Database=ScrantonPaperCo ; Trusted_Connection=Yes;"
```

```python
query = 'SELECT * FROM Performance_Reviews'
```

```python
# using with context manager
with pyodbc.connect(conn_string) as conx:
    cursor = conx.cursor()
    cursor.execute(query)
    data = cursor.fetchall()

# Display top 5 rows of the result
for each in data[:5]:
    print(each)
```

```python
# Display all tables in the dataframe
tables = [row.table_name for row in cursor.tables()]
print(tables[:3])
```
